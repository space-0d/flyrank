## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from google.colab import files
import pandas as pd
import io

# Upload the CSV
uploaded = files.upload()

# Read the uploaded CSV
filename = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("File loaded:", filename)
print("Rows:", len(df))
print("Columns:", len(df.columns))

# Quick look at the data
print("\nQuick look at the data:")

print("Total content pages:", len(df))

if "content_age_days" in df.columns:
    print(
        "Median content age:",
        round(df["content_age_days"].median(), 1),
        "days"
    )

if "days_since_last_update" in df.columns:
    print(
        "Median days since last update:",
        round(df["days_since_last_update"].median(), 1),
        "days"
    )

if "impressions_90d" in df.columns:
    median_impressions = df["impressions_90d"].median()
    print(
        "Median 90-day impressions:",
        round(median_impressions, 1)
    )

if "trend_direction" in df.columns:
    declining = df["trend_direction"].eq("down")
    print(
        "Declining pages:",
        declining.sum()
    )
    print(
        "Declining percentage:",
        round(declining.mean() * 100, 1),
        "%"
    )
```
